# Package-first quickstart
This notebook demonstrates the same API available to scripts, services and the CLI.


```python
from sklearn.datasets import load_iris
from ml_unsupervised import ClusteringPipeline
```

```python
X = load_iris().data
pipeline = ClusteringPipeline(
    scaler="standard",
    reduction_method="pca",
    n_components=2,
    clustering_method="kmeans",
    clustering_params={"n_clusters": 3, "random_state": 42, "n_init": 10},
)
result = pipeline.fit_predict(X)
result.metrics.to_dict() if result.metrics else None
```
